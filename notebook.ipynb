# Uso de matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([[255, 0, 90], [0, 90, 0]])
x
```

```python
plt.figure(figsize=(5, 5))
plt.subplot(2,2,1)
plt.imshow(x, vmin=0, vmax=255, cmap='gray')
plt.title('Escala de grises')
plt.subplot(2,2,2)
plt.imshow(x, vmin=0, vmax=255, cmap='viridis')
plt.title('Viridis')
plt.subplot(2,2,3)
plt.imshow(x, vmin=0, vmax=255, cmap='magma')
plt.title('Magma')
```

# Laboratorio: Comparación de Mapas de Color

En este laboratorio, exploraremos cómo diferentes mapas de color (`cmap`) afectan la visualización de imágenes en escala de grises. Esto es fundamental para seleccionar el mapa adecuado en aplicaciones científicas y de visión por computador.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Crear una imagen sintética en escala de grises
gray_image = np.linspace(0, 255, 256).reshape(16, 16).astype(np.uint8)

# Visualizar la imagen con diferentes mapas de color
colormaps = ['gray', 'viridis', 'plasma', 'inferno', 'magma']

plt.figure(figsize=(15, 5))
for i, cmap in enumerate(colormaps):
    plt.subplot(1, len(colormaps), i + 1)
    plt.imshow(gray_image, cmap=cmap)
    plt.title(cmap)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
